```python
# sabse pahle  age column me mean se replace karna hoga missing values ko
# cabin column me most frequent se replace karna hoga  missing values ko
# embarked columns me jo missing rows hai usko remove karna hoga
```

```python
from sklearn.impute import SimpleImputer
si_age=SimpleImputer(missing_values=np.nan,strategy='mean')
df[['Age']]=si_age.fit_transform(df[['Age']])
```

```python
df['Age'].isnull().sum()
```

```python
si_cabin=SimpleImputer(missing_values=np.nan,strategy='most_frequent')
df[['Cabin']]=si_cabin.fit_transform(df[['Cabin']])
```

```python
df['Cabin'].isnull().sum()
```

```python
df.dropna(subset=['Embarked'],inplace=True,axis=0)
df.reset_index(drop=True,inplace=True)
```

```python
df['Embarked'].isnull().sum()
```

```python
# categorical encoding karna hoga saare categorical columns pe
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
categorical_cols=['Sex','Embarked',]
trf=ColumnTransformer([
    ('trf',OneHotEncoder(drop='first',handle_unknown='ignore'),categorical_cols)
],remainder='passthrough')
```

```python
# mere data set me passenger id  name and Cabin ka jada kaam nahi hai toh mai usko drop karne ka soha hu
```

```python
df.drop(columns=['PassengerId','Name','Cabin'])
```

```python
# training and testing data ko alag alag karna 
```

```python
X=df.drop(columns=['PassengerId','Name','Cabin','Ticket','Survived'])
Y=df['Survived']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train
```

```python
Y_train
```

```python
# column transformer ko train data par apply karna
X_train_trf=trf.fit_transform(X_train)
X_test_trf=trf.transform(X_test)
```

```python
# ml algo apply kiya
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
lr=LogisticRegression()
lr.fit(X_train_trf,Y_train)
y_pred=lr.predict(X_test_trf)
print('LogisticRegression Accuracy:',accuracy_score(Y_test,y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc_model=RandomForestClassifier(random_state=42)
rfc_model.fit(X_train_trf,Y_train)
y_pred_rfc=rfc_model.predict(X_test_trf)
print("RandomForestClassifier Score:",accuracy_score(Y_test,y_pred_rfc))
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc_model=DecisionTreeClassifier()
dtc_model.fit(X_train_trf,Y_train)
y_pred_dtc=dtc_model.predict(X_test_trf)
print("DecisionTreeClassifier Score:",accuracy_score(Y_test,y_pred_dtc))
```

```python
# maine teen ml algo check kiya hai o r mujhe sabse aacha accuracy LogisticRegression me mila
```

```python
# saving the model
```

```python
import pickle

# 1. Logistic Regression model ko save karna
with open('lr_model.pkl', 'wb') as file:
    pickle.dump(lr, file)

# 2. ColumnTransformer ko bhi save karna zaroori hai (taaki naye data par preprocessing apply ho sake)
with open('transformer.pkl', 'wb') as file:
    pickle.dump(trf, file)

print("Model and Transformer successfully saved!")
```
